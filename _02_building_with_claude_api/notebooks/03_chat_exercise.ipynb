What is required as part of this notebook exercise?!

1. Create a chat like program which takes user input and replies with assistant response. 
2. It read input using built-in input() function
3. Calls Anthropic api
4. Prints llm response
5. Maintains chat history - gives a chatroom type feel

```python
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()
```

```python
#define required parameters
model = "claude-sonnet-5"
messages = []

client = Anthropic()
```

```python
#hepler functions
def add_user_message(messages, text):
    message = {"role": "user", "content": text}
    messages.append(message)

def add_assistant_message(messages, text):
    message = {"role": "assistant", "content": text}
    messages.append(message)

def chat(messages):
    response = client.messages.create(
        model=model,
        max_tokens=256,
        messages=messages
    )
    # content is a list of blocks -- not guaranteed to be all text
    # (e.g. a ThinkingBlock has .thinking, not .text). Pull out only
    # the actual text block(s), ignoring anything else.
    text_blocks = [block.text for block in response.content if block.type == "text"]
    return "\n".join(text_blocks)

def chatroom():
    user_input = input()
    print("> " + user_input)

    #creates user message
    add_user_message(messages, user_input)

    #call model
    resp_text = chat(messages)


    #appends to messages
    add_assistant_message(messages, resp_text)

    #prints assistant response
    print("---")
    print(resp_text)
    print("---")
```

```python
print("====[Simple Chat App]====")
while True:
    chatroom()
```
