```python
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()
```

## The client

The `Anthropic` client is what holds the API key and handles auth (the `x-api-key` header) for every request you make with it — this is the SDK's version of leg 2 from the notes.

```python
client = Anthropic()
model = "claude-sonnet-5"
```

## The first request

Same four things from the notes: `model`, `messages` (the input), `max_tokens` (cap on *this response's* output length -- not the context window).

```python
message = client.messages.create(
    model = model,
    max_tokens = 500,
    messages = [
        {
            "role": "user",
            "content": "What is cloud computing? Answer in one sentence."
        }
    ]
)
```

```python
print(message)
```

## Inspecting the response

Matches the response anatomy from the notes: `content` (the generated message), `usage` (input/output token counts), `stop_reason` (why generation stopped).

```python
print("Message:", message.content[0].text)
print("Stop reason:", message.stop_reason)
print("Usage:", message.usage)
```
